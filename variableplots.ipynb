```julia
using CSV
using DataFrames
using Statistics

data = CSV.read("Location4.csv", DataFrame)
sort!(data, :Time)

features = [
    :temperature_2m,
    :relativehumidity_2m,
    :dewpoint_2m,
    :windspeed_10m,
    :windspeed_100m,
    :winddirection_10m,
    :winddirection_100m,
    :windgusts_10m,
    :Power
]

# Calculate hourly differences
delta_data = DataFrame()

for feature in features
    delta_data[!, feature] = diff(Float64.(data[!, feature]))
end

# Correlation of each feature's change with the change in power
correlations = [
    cor(
        delta_data[!, feature],
        delta_data.Power
    )
    for feature in features
]

correlation_results = DataFrame(
    Feature = string.(features),
    CorrelationWithDeltaPower = correlations
)

sort!(correlation_results, :CorrelationWithDeltaPower, by=abs, rev=true)

display(correlation_results)
```

```julia
using CSV
using DataFrames
using Dates
using Statistics
using Plots

data = CSV.read("Location4.csv", DataFrame)
data.Time = DateTime.(data.Time)
sort!(data, :Time)

# Calculate hourly differences
data.delta_power = [missing; diff(Float64.(data.Power))]
data.delta_wind10 = [missing; diff(Float64.(data.windspeed_10m))]
data.delta_wind100 = [missing; diff(Float64.(data.windspeed_100m))]
data.delta_gust = [missing; diff(Float64.(data.windgusts_10m))]

wind_features = [
    (:delta_wind10, "Wind speed 10 m"),
    (:delta_wind100, "Wind speed 100 m"),
    (:delta_gust, "Wind gust")
]

results = DataFrame(
    Feature = String[],
    LagHours = Int[],
    Correlation = Float64[],
    SampleCount = Int[]
)

for (feature, label) in wind_features
    for lag in 0:24
        wind_changes = Float64[]
        power_changes = Float64[]

        # i is the time of the power change being predicted
        for i in (lag + 2):nrow(data)
            # Ensure all timestamps from i-lag-1 through i are hourly
            if all(
                data.Time[j + 1] - data.Time[j] == Hour(1)
                for j in (i - lag - 1):(i - 1)
            )
                w = data[i - lag, feature]
                p = data[i, :delta_power]

                if !ismissing(w) && !ismissing(p) &&
                   isfinite(w) && isfinite(p)
                    push!(wind_changes, w)
                    push!(power_changes, p)
                end
            end
        end

        r = length(wind_changes) >= 3 ?
            cor(wind_changes, power_changes) : NaN

        push!(results, (
            label, lag, r, length(wind_changes)
        ))
    end
end

display(results)

# Plot correlation against lag for each wind feature
p = plot(
    xlabel = "Wind change lead time (hours)",
    ylabel = "Correlation with future power change",
    title = "Lagged Wind–Power Change Correlations",
    ylims = (-1, 1),
    legend = :best
)

for (feature, label) in wind_features
    subset = filter(row -> row.Feature == label, results)
    plot!(p, subset.LagHours, subset.Correlation,
        marker = :circle, label = label)
end

display(p)
```

```julia

using CSV
using DataFrames
using Dates
using Statistics
using Random
using LinearAlgebra
using Plots

purple = RGB(122/256, 30/256, 71/256)
blue = RGB(0/256, 39/256, 76/256)
mygreen = RGB(83/256, 165/256, 132/256)
orange = RGB(220/256, 120/256, 40/256)
location_colors = [purple, blue, mygreen, orange]

# Fixed hyperparameters
alpha = 0.001
iter_max = 25_000
lambda = 0.001
gamma = 5.0
centers = collect(0.0:0.05:1.1)

# --------------------------------------------------
# Load and combine all four locations
# --------------------------------------------------

data = DataFrame()

for loc in 1:4
    filename = "Location$(loc).csv"
    location_data = CSV.read(filename, DataFrame)

    location_data.Time = DateTime.(location_data.Time)
    location_data.LocationID = fill(loc, nrow(location_data))

    append!(data, location_data; cols=:union)
end

sort!(data, [:LocationID, :Time])

n = nrow(data)
power = Float64.(data.Power)
x10 = Float64.(data.windspeed_10m) ./ 10.0
x100 = Float64.(data.windspeed_100m) ./ 10.0
xgust = Float64.(data.windgusts_10m) ./ 10.0

location_ids = Int.(data.LocationID)
day = Date.(data.Time)

# --------------------------------------------------
# Identify the previous 1, 2, and 3 hourly rows
# --------------------------------------------------

prev_idx = zeros(Int, n)
prev2_idx = zeros(Int, n)
prev3_idx = zeros(Int, n)

for i in 2:n
    if location_ids[i] == location_ids[i-1] &&
       data.Time[i] - data.Time[i-1] == Hour(1)

        prev_idx[i] = i - 1

        if prev_idx[i-1] > 0
            prev2_idx[i] = prev_idx[i-1]

            if prev2_idx[i-1] > 0
                prev3_idx[i] = prev2_idx[i-1]
            end
        end
    end
end

# --------------------------------------------------
# Calculate power and weather changes
# --------------------------------------------------

delta_power = fill(NaN, n)
delta_x10 = fill(NaN, n)
delta_x100 = fill(NaN, n)
delta_xgust = fill(NaN, n)

for i in 1:n
    if prev_idx[i] > 0
        j = prev_idx[i]

        delta_power[i] = power[i] - power[j]
        delta_x10[i] = x10[i] - x10[j]
        delta_x100[i] = x100[i] - x100[j]
        delta_xgust[i] = xgust[i] - xgust[j]
    end
end

# --------------------------------------------------
# Construct the model's extra features
#
# Columns 1-3:  current weather changes
# Columns 4-6:  weather changes at lag 1
# Columns 7-9:  weather changes at lag 2
# Column 10:    power change at lag 1
# Column 11:    power level at lag 1
# --------------------------------------------------

raw_extra = fill(NaN, n, 11)

for i in 1:n
    if prev_idx[i] > 0 &&
       prev2_idx[i] > 0 &&
       prev3_idx[i] > 0

        j1 = prev_idx[i]
        j2 = prev2_idx[i]

        raw_extra[i, :] = [
            delta_x10[i],
            delta_x100[i],
            delta_xgust[i],

            delta_x10[j1],
            delta_x100[j1],
            delta_xgust[j1],

            delta_x10[j2],
            delta_x100[j2],
            delta_xgust[j2],

            delta_power[j1],
            power[j1]
        ]
    end
end

# Current wind levels are used to construct the RBF features.
Xrbf = hcat(x10, x100, xgust)

valid = falses(n)

for i in 1:n
    valid[i] =
        prev3_idx[i] > 0 &&
        isfinite(delta_power[i]) &&
        all(isfinite, raw_extra[i, :]) &&
        all(isfinite, Xrbf[i, :])
end

# --------------------------------------------------
# Identify complete, consecutive 24-hour days
# for all four locations
# --------------------------------------------------

all_dates = sort(unique(day))
all_locations = sort(unique(location_ids))
eligible_days = Date[]

for d in all_dates
    complete_day = true

    for loc in all_locations
        rows = findall((day .== d) .& (location_ids .== loc))

        if length(rows) != 24
            complete_day = false
            break
        end

        if hour.(data.Time[rows]) != collect(0:23)
            complete_day = false
            break
        end

        if !all(valid[rows])
            complete_day = false
            break
        end
    end

    if complete_day
        push!(eligible_days, d)
    end
end

if length(eligible_days) < 3
    error("Not enough complete days to create training, validation, and test sets.")
end

# Latest complete day is the test day.
test_day = last(eligible_days)

# Randomly select up to 14 validation days before the test day.
validation_candidates = eligible_days[1:end-1]
n_validation_days = min(14, length(validation_candidates))

validation_days = sort(
    Random.shuffle(validation_candidates)[1:n_validation_days]
)

# Training uses days strictly before the earliest validation day.
train_days = filter(d -> d < minimum(validation_days), eligible_days)

if isempty(train_days)
    error("No training days remain. Use fewer validation days.")
end

train_idx = findall(
    i -> valid[i] && day[i] in train_days,
    1:n
)

validation_idx = findall(
    i -> valid[i] && day[i] in validation_days,
    1:n
)

test_idx = findall(
    i -> valid[i] && day[i] == test_day,
    1:n
)

println("Training days: ", length(train_days))
println("Validation days: ", length(validation_days))
println("Test day: ", test_day)
println("Training observations: ", length(train_idx))
println("Validation observations: ", length(validation_idx))
println("Test observations: ", length(test_idx))

# --------------------------------------------------
# Standardize the extra features using training data
# only, preventing validation/test leakage
# --------------------------------------------------

extra_mean = vec(mean(raw_extra[train_idx, :], dims=1))
extra_scale = vec(std(raw_extra[train_idx, :], dims=1))

extra_scale[extra_scale .< eps(Float64)] .= 1.0

extra_std = (
    raw_extra .- reshape(extra_mean, 1, length(extra_mean))
) ./ reshape(extra_scale, 1, length(extra_scale))

# Standardize the target: change in power.
target_mean = mean(delta_power[train_idx])
target_scale = std(delta_power[train_idx])

if !isfinite(target_scale) || target_scale < eps(Float64)
    target_scale = 1.0
end

target_std = (delta_power .- target_mean) ./ target_scale

# --------------------------------------------------
# Gaussian RBF features
# --------------------------------------------------

function rbf_features(x, centers, gamma)
    return vcat(
        (
            exp.(-gamma .* (x[j] .- centers).^2)
            for j in eachindex(x)
        )...
    )
end

rbf_matrix = zeros(n, 3 * length(centers))

for i in 1:n
    rbf_matrix[i, :] = rbf_features(
        vec(Xrbf[i, :]),
        centers,
        gamma
    )
end

# --------------------------------------------------
# Train the RBF + linear-feature model using SGD
# --------------------------------------------------

function fit_rbf(
    train_idx,
    rbf_matrix,
    extra_std,
    target_std,
    alpha,
    iter_max,
    lambda
)
    n_rbf = size(rbf_matrix, 2)
    n_extra = size(extra_std, 2)

    # Final coefficient is the intercept.
    weights = zeros(n_rbf + n_extra + 1)

    for iteration in 1:iter_max
        i = rand(train_idx)

        xi = vcat(
            rbf_matrix[i, :],
            extra_std[i, :],
            1.0
        )

        prediction = dot(weights, xi)
        error = prediction - target_std[i]

        gradient = 2.0 .* error .* xi

        # L2 regularization, excluding the intercept.
        gradient[1:end-1] .+=
            2.0 * lambda .* weights[1:end-1]

        weights .-= alpha .* gradient
    end

    return weights
end

weights = fit_rbf(
    train_idx,
    rbf_matrix,
    extra_std,
    target_std,
    alpha,
    iter_max,
    lambda
)

# --------------------------------------------------
# Predict a power change
#
# During recursive forecasts, columns 10 and 11
# use the model's predicted lagged power values.
# --------------------------------------------------

function predict_delta(
    i,
    previous_power,
    previous_delta_power,
    rbf_matrix,
    extra_std,
    weights,
    extra_mean,
    extra_scale,
    target_mean,
    target_scale
)
    zextra = copy(extra_std[i, :])

    # Column 10: previous power change.
    zextra[10] =
        (previous_delta_power - extra_mean[10]) /
        extra_scale[10]

    # Column 11: previous power level.
    zextra[11] =
        (previous_power - extra_mean[11]) /
        extra_scale[11]

    xi = vcat(
        rbf_matrix[i, :],
        zextra,
        1.0
    )

    prediction_std = dot(weights, xi)

    return prediction_std * target_scale + target_mean
end

# --------------------------------------------------
# Training RMSE: one-step-ahead predictions using
# actual previous power values
# --------------------------------------------------

train_predictions = Float64[]

for i in train_idx
    j = prev_idx[i]
    previous_power = power[j]
    previous_delta = delta_power[j]

    predicted_delta = predict_delta(
        i,
        previous_power,
        previous_delta,
        rbf_matrix,
        extra_std,
        weights,
        extra_mean,
        extra_scale,
        target_mean,
        target_scale
    )

    push!(train_predictions, previous_power + predicted_delta)
end

train_rmse = sqrt(mean(
    (power[train_idx] .- train_predictions).^2
))

# --------------------------------------------------
# Recursively forecast a complete day for one location
# --------------------------------------------------

function recursive_forecast(
    rows,
    data,
    power,
    delta_power,
    prev_idx,
    rbf_matrix,
    extra_std,
    weights,
    extra_mean,
    extra_scale,
    target_mean,
    target_scale
)
    first_row = first(rows)
    previous_row = prev_idx[first_row]

    if previous_row == 0 || prev_idx[previous_row] == 0
        error("Insufficient historical power data before forecast day.")
    end

    previous_power = power[previous_row]
    previous_delta = delta_power[previous_row]

    predicted_power = Float64[]
    predicted_delta = Float64[]

    for i in rows
        delta_hat = predict_delta(
            i,
            previous_power,
            previous_delta,
            rbf_matrix,
            extra_std,
            weights,
            extra_mean,
            extra_scale,
            target_mean,
            target_scale
        )

        power_hat = previous_power + delta_hat

        push!(predicted_delta, delta_hat)
        push!(predicted_power, power_hat)

        # Feed predictions back into the next hour.
        previous_delta = delta_hat
        previous_power = power_hat
    end

    return predicted_power, predicted_delta
end

# --------------------------------------------------
# Forecast selected days for every location
# --------------------------------------------------

function forecast_days(
    selected_days,
    data,
    day,
    location_ids,
    valid,
    power,
    delta_power,
    prev_idx,
    rbf_matrix,
    extra_std,
    weights,
    extra_mean,
    extra_scale,
    target_mean,
    target_scale
)
    result = DataFrame(
        Time = DateTime[],
        LocationID = Int[],
        ActualPower = Float64[],
        PredictedPower = Float64[],
        ActualDeltaPower = Float64[],
        PredictedDeltaPower = Float64[]
    )

    for d in selected_days
        for loc in sort(unique(location_ids))
            rows = findall(
                (day .== d) .&
                (location_ids .== loc) .&
                valid
            )

            isempty(rows) && continue

            sort!(rows, by=i -> data.Time[i])

            predicted_power, predicted_delta = recursive_forecast(
                rows,
                data,
                power,
                delta_power,
                prev_idx,
                rbf_matrix,
                extra_std,
                weights,
                extra_mean,
                extra_scale,
                target_mean,
                target_scale
            )

            for (k, i) in enumerate(rows)
                push!(result, (
                    data.Time[i],
                    location_ids[i],
                    power[i],
                    predicted_power[k],
                    delta_power[i],
                    predicted_delta[k]
                ))
            end
        end
    end

    sort!(result, [:LocationID, :Time])

    return result
end

validation_results = forecast_days(
    validation_days,
    data,
    day,
    location_ids,
    valid,
    power,
    delta_power,
    prev_idx,
    rbf_matrix,
    extra_std,
    weights,
    extra_mean,
    extra_scale,
    target_mean,
    target_scale
)

test_results = forecast_days(
    [test_day],
    data,
    day,
    location_ids,
    valid,
    power,
    delta_power,
    prev_idx,
    rbf_matrix,
    extra_std,
    weights,
    extra_mean,
    extra_scale,
    target_mean,
    target_scale
)

validation_rmse = sqrt(mean(
    (validation_results.ActualPower .-
     validation_results.PredictedPower).^2
))

test_rmse = sqrt(mean(
    (test_results.ActualPower .-
     test_results.PredictedPower).^2
))

println("\nTraining RMSE: ", train_rmse)
println("Validation RMSE: ", validation_rmse)
println("Test RMSE: ", test_rmse)

CSV.write("validation_predictions.csv", validation_results)
CSV.write("test_predictions.csv", test_results)



# --------------------------------------------------
# Lag correlation helper
#
# corr(a[t], b[t-lag]), calculated within each
# location and only across consecutive hourly rows
# --------------------------------------------------

function lag_correlation(
    a,
    b,
    lag,
    prev_idx,
    valid,
    location_ids
)
    a_values = Float64[]
    b_values = Float64[]

    for i in eachindex(a)
        valid[i] || continue

        j = i

        for _ in 1:lag
            if j == 0 || prev_idx[j] == 0
                j = 0
                break
            end

            j = prev_idx[j]
        end

        if j > 0 &&
           valid[j] &&
           location_ids[i] == location_ids[j] &&
           isfinite(a[i]) &&
           isfinite(b[j])

            push!(a_values, a[i])
            push!(b_values, b[j])
        end
    end

    if length(a_values) < 3 ||
       std(a_values) == 0 ||
       std(b_values) == 0
        return NaN
    end

    return cor(a_values, b_values)
end

# --------------------------------------------------
# Autocorrelation of power changes, lags 1-24
# --------------------------------------------------

power_acf = [
    lag_correlation(
        delta_power,
        delta_power,
        lag,
        prev_idx,
        valid,
        location_ids
    )
    for lag in 1:24
]

p_acf = plot(
    1:24,
    power_acf,
    marker = :circle,
    color = purple,
    xlabel = "Lag (hours)",
    ylabel = "Correlation",
    title = "Autocorrelation of Power Changes",
    legend = false,
    ylims = (-1, 1)
)

display(p_acf)
```
